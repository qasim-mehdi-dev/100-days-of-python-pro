### Convert Strings to DateTime Objects

```python
print(f"Tesla date type: {df_tesla['MONTH'].dtype}")
print(f"Unemployment date type: {df_unemployment['MONTH'].dtype}")
print(f"BTC Search date type: {df_btc_search['MONTH'].dtype}")
print(f"BTC Price date type: {df_btc_price['DATE'].dtype}")
```

```python
print(f"Fixed Tesla type: {df_tesla['MONTH'].dtype}")
print(f"Fixed BTC Price type: {df_btc_price['DATE'].dtype}")
```

### Converting from Daily to Monthly Data



```python
df_btc_monthly = df_btc_price.resample('ME', on='DATE').mean()

print(f"New monthly BTC shape: {df_btc_monthly.shape}")
df_btc_monthly.head()
```

### Tesla Stock Price v.s. Search Volume

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# 1. Figure setup
plt.figure(figsize=(14, 8), dpi=120)
plt.title('Tesla Stock Price vs Search Trend', fontsize=18)

# 2. Get the current axes (ax1) and plot Search Trend (Blue)
ax1 = plt.gca() 
ax2 = ax1.twinx() # Create a twin y-axis sharing the same x-axis

ax1.plot(df_tesla['MONTH'], df_tesla['TSLA_WEB_SEARCH'], color='skyblue', linewidth=3)
ax2.plot(df_tesla['MONTH'], df_tesla['TSLA_USD_CLOSE'], color='#E63946', linewidth=3)

# 3. Label the axes
ax1.set_ylabel('Search Trend', color='skyblue', fontsize=14)
ax1.set_xlabel('Month', fontsize=14)
ax2.set_ylabel('TSLA Stock Price', color='#E63946', fontsize=14)

# 5. Format the x-axis ticks
years = mdates.YearLocator()
months = mdates.MonthLocator()
years_fmt = mdates.DateFormatter('%Y')

ax1.xaxis.set_major_locator(years)
ax1.xaxis.set_major_formatter(years_fmt)
ax1.xaxis.set_minor_locator(months)

# 4. Display the chart
plt.show()
```

### Bitcoin (BTC) Price v.s. Search Volume

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# 1. Figure setup with crisp DPI
plt.figure(figsize=(14, 8), dpi=120)

# Challenge #1: Modify the chart title
plt.title('Bitcoin News Search vs Resampled Price', fontsize=18)

# Axis setup
ax1 = plt.gca()
ax2 = ax1.twinx()

# Plot Search (Skyblue) with Challenge #5: circular markers
ax1.plot(df_btc_search['MONTH'], df_btc_search['BTC_NEWS_SEARCH'], 
         color='skyblue', linewidth=3, marker='o')

# Plot Price (Orange/Red) with Challenge #4: dashed line style
ax2.plot(df_btc_monthly.index, df_btc_monthly['CLOSE'], 
         color='#F7931A', linewidth=3, linestyle='--')

# Labels and Challenge #2: Change y-axis label to 'BTC Price'
ax1.set_ylabel('Search Trend', color='skyblue', fontsize=14)
ax1.set_xlabel('Month', fontsize=14)
ax2.set_ylabel('BTC Price', color='#F7931A', fontsize=14)

# Tick formatting and rotation
ax1.tick_params(axis='x', labelsize=14, rotation=45)
ax1.tick_params(axis='y', labelsize=14)
ax2.tick_params(axis='y', labelsize=14)

# Time axis locator configuration
years = mdates.YearLocator()
months = mdates.MonthLocator()
years_fmt = mdates.DateFormatter('%Y')
ax1.xaxis.set_major_locator(years)
ax1.xaxis.set_major_formatter(years_fmt)
ax1.xaxis.set_minor_locator(months)

# Challenge #3: Change limits to improve appearance and remove empty space
ax1.set_xlim([df_btc_search['MONTH'].min(), df_btc_search['MONTH'].max()])
ax1.set_ylim([0, df_btc_search['BTC_NEWS_SEARCH'].max() * 1.05])
ax2.set_ylim([0, df_btc_monthly['CLOSE'].max() * 1.05])

# Explicit render
plt.show()
```

### Unemployement Benefits Search vs. Actual Unemployment in the U.S.

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# 1. Figure setup with crisp DPI
plt.figure(figsize=(14, 8), dpi=120)

# Challenge #1: Change the title
plt.title('Monthly Search of "Unemployment Benefits" in the U.S. vs the U/E Rate', fontsize=18)

# Axis setup
ax1 = plt.gca()
ax2 = ax1.twinx()

# Challenge #4: Add a grey dashed grid onto the background axis
ax1.grid(color='grey', linestyle='--')

# Plot search data on ax1 and official U/E rate on ax2
ax1.plot(df_unemployment['MONTH'], df_unemployment['UE_BENEFITS_WEB_SEARCH'], color='skyblue', linewidth=3)
ax2.plot(df_unemployment['MONTH'], df_unemployment['UNRATE'], color='#E63946', linewidth=3)

# Challenge #2: Change y-axis labels (Note: Right side is FRED U/E Rate)
ax1.set_ylabel('Search Trend', color='skyblue', fontsize=14)
ax1.set_xlabel('Month', fontsize=14)
ax2.set_ylabel('FRED U/E Rate', color='#E63946', fontsize=14)

# Tick formatting and rotation
ax1.tick_params(axis='x', labelsize=14, rotation=45)
ax1.tick_params(axis='y', labelsize=14)
ax2.tick_params(axis='y', labelsize=14)

# Time axis locator configuration
years = mdates.YearLocator()
months = mdates.MonthLocator()
years_fmt = mdates.DateFormatter('%Y')
ax1.xaxis.set_major_locator(years)
ax1.xaxis.set_major_formatter(years_fmt)
ax1.xaxis.set_minor_locator(months)

# Challenge #3: Change the axis limits to trim empty whitespace padding cleanly
ax1.set_xlim([df_unemployment['MONTH'].min(), df_unemployment['MONTH'].max()])
ax1.set_ylim([0, df_unemployment['UE_BENEFITS_WEB_SEARCH'].max() * 1.05])
ax2.set_ylim([0, df_unemployment['UNRATE'].max() * 1.05])

# Explicit render
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# 1. Calculate the rolling average for the search metrics
roll_df = df_unemployment['UE_BENEFITS_WEB_SEARCH'].rolling(window=6).mean()

# 2. Figure layout
plt.figure(figsize=(14, 8), dpi=120)
plt.title('Rolling 6-Month Average of "Unemployment Benefits" Searches vs U/E Rate', fontsize=18)

ax1 = plt.gca()
ax2 = ax1.twinx()

ax1.grid(color='grey', linestyle='--')

# 3. Plot the rolling smooth search line (ax1) and actual rate (ax2)
ax1.plot(df_unemployment['MONTH'], roll_df, color='skyblue', linewidth=3)
ax2.plot(df_unemployment['MONTH'], df_unemployment['UNRATE'], color='#E63946', linewidth=3)

# 4. Labels and clean bounds
ax1.set_ylabel('Search Trend (6mo Rolling)', color='skyblue', fontsize=14)
ax1.set_xlabel('Month', fontsize=14)
ax2.set_ylabel('FRED U/E Rate', color='#E63946', fontsize=14)

ax1.tick_params(axis='x', labelsize=14, rotation=45)
ax1.tick_params(axis='y', labelsize=14)
ax2.tick_params(axis='y', labelsize=14)

years = mdates.YearLocator()
months = mdates.MonthLocator()
years_fmt = mdates.DateFormatter('%Y')
ax1.xaxis.set_major_locator(years)
ax1.xaxis.set_major_formatter(years_fmt)
ax1.xaxis.set_minor_locator(months)

ax1.set_xlim([df_unemployment['MONTH'].min(), df_unemployment['MONTH'].max()])
ax1.set_ylim([0, roll_df.max() * 1.05])
ax2.set_ylim([0, df_unemployment['UNRATE'].max() * 1.05])

plt.show()
```

### 2020 in Unemployment Charts

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# 1. Read the new CSV data into a DataFrame
df_ue_2020 = pd.read_csv('UE Benefits Search vs UE Rate 2004-20.csv')

# 2. Convert the MONTH column to Pandas Datetime objects
df_ue_2020['MONTH'] = pd.to_datetime(df_ue_2020['MONTH'])

# 3. Figure setup with high sharpness
plt.figure(figsize=(14, 8), dpi=120)
plt.title('Monthly Search of "Unemployment Benefits" vs U/E Rate (2004-2020)', fontsize=18)

ax1 = plt.gca()
ax2 = ax1.twinx()

# Add our trusty background grid
ax1.grid(color='grey', linestyle='--')

# 4. Plot the data strings
ax1.plot(df_ue_2020['MONTH'], df_ue_2020['UE_BENEFITS_WEB_SEARCH'], color='skyblue', linewidth=3)
ax2.plot(df_ue_2020['MONTH'], df_ue_2020['UNRATE'], color='#E63946', linewidth=3)

# 5. Labels and styling
ax1.set_ylabel('Search Trend', color='skyblue', fontsize=14)
ax1.set_xlabel('Month', fontsize=14)
ax2.set_ylabel('FRED U/E Rate', color='#E63946', fontsize=14)

ax1.tick_params(axis='x', labelsize=14, rotation=45)
ax1.tick_params(axis='y', labelsize=14)
ax2.tick_params(axis='y', labelsize=14)

# Format time markers
years = mdates.YearLocator()
months = mdates.MonthLocator()
years_fmt = mdates.DateFormatter('%Y')
ax1.xaxis.set_major_locator(years)
ax1.xaxis.set_major_formatter(years_fmt)
ax1.xaxis.set_minor_locator(months)

# Fit snug boundaries
ax1.set_xlim([df_ue_2020['MONTH'].min(), df_ue_2020['MONTH'].max()])
ax1.set_ylim([0, df_ue_2020['UE_BENEFITS_WEB_SEARCH'].max() * 1.05])
ax2.set_ylim([0, df_ue_2020['UNRATE'].max() * 1.05])

# Render the visualization
plt.show()
```
