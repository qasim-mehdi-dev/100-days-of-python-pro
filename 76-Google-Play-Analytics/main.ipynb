# Import Statements

```python
import pandas as pd
import plotly.express as px
```

# Read the Dataset

```python
df_apps = pd.read_csv('apps.csv')
```

# Data Cleaning

```python
df_apps.shape
```

```python
df_apps.head()
```

```python
df_apps.sample(5)
```

```python
df_apps.drop(['Last_Updated', 'Android_Ver'], axis=1, inplace=True)
df_apps.head()
```

```python
nan_rows = df_apps[df_apps.Rating.isna()]
print(nan_rows.shape)
nan_rows.head()
```

```python
df_apps_clean = df_apps.dropna()
df_apps_clean.shape
```

```python
duplicated_rows = df_apps_clean[df_apps_clean.duplicated()]
print(duplicated_rows.shape)
duplicated_rows.head()
```

```python
df_apps_clean[df_apps_clean.App == 'Instagram']
```

```python
df_apps_clean = df_apps_clean.drop_duplicates()
```

```python
df_apps_clean[df_apps_clean.App == 'Instagram']
```

```python
df_apps_clean = df_apps_clean.drop_duplicates(subset=['App', 'Type', 'Price'])
df_apps_clean[df_apps_clean.App == 'Instagram']
```

```python
df_apps_clean.shape
```

```python
df_apps_clean.sort_values('Rating', ascending=False).head()
```

```python
df_apps_clean.sort_values('Size_MBs', ascending=False).head()
```

```python
df_apps_clean.sort_values('Reviews', ascending=False).head(50)
```

```python
ratings = df_apps_clean.Content_Rating.value_counts()
ratings
```

```python
fig = px.pie(labels=ratings.index, values=ratings.values)

fig.show()
```

```python
fig = px.pie(labels=ratings.index, 
             values=ratings.values,
             title="Content Rating",
             names=ratings.index,
)
fig.update_traces(textposition='outside', textinfo='percent+label')

fig.show()
```

```python
fig = px.pie(labels=ratings.index, 
             values=ratings.values,
             title="Content Rating",
             names=ratings.index,
             hole=0.6, 
)
fig.update_traces(textposition='inside', textfont_size=15, textinfo='percent')

fig.show()
```

```python
df_apps_clean.Installs.describe()
```

```python
df_apps_clean.info()
```

```python
df_apps_clean[['App', 'Installs']].groupby('Installs').count()
```

```python
df_apps_clean.Installs = df_apps_clean.Installs.astype(str).str.replace(',', "")

df_apps_clean.Installs = pd.to_numeric(df_apps_clean.Installs)

df_apps_clean[['App', 'Installs']].groupby('Installs').count()
```

```python
df_apps_clean.Price = df_apps_clean.Price.astype(str).str.replace('$', "")
df_apps_clean.Price = pd.to_numeric(df_apps_clean.Price)

df_apps_clean.sort_values('Price', ascending=False).head(20)
```

```python
df_apps_clean = df_apps_clean[df_apps_clean['Price'] < 250]
df_apps_clean.sort_values('Price', ascending=False).head(5)
```

```python
df_apps_clean['Revenue_Estimate'] = df_apps_clean.Installs.mul(df_apps_clean.Price)
df_apps_clean.sort_values('Revenue_Estimate', ascending=False)[:10]
```

```python
df_apps_clean.Category.nunique()
```

```python
top10_category = df_apps_clean.Category.value_counts()[:10]
top10_category
```

```python
bar = px.bar(
        x = top10_category.index, 
        y = top10_category.values)

bar.show()
```

```python
category_installs = df_apps_clean.groupby('Category').agg({'Installs': pd.Series.sum})
category_installs.sort_values('Installs', ascending=True, inplace=True)
```

```python
h_bar = px.bar(
        x = category_installs.Installs,
        y = category_installs.index,
        orientation='h',
        title='Category Popularity')

h_bar.update_layout(xaxis_title='Number of Downloads', yaxis_title='Category')
h_bar.show()
```

```python
cat_number = df_apps_clean.groupby('Category').agg({'App': pd.Series.count})
cat_merged_df = pd.merge(cat_number, category_installs, on='Category', how="inner")
print(f'The dimensions of the DataFrame are: {cat_merged_df.shape}')
cat_merged_df.sort_values('Installs', ascending=False)
```

```python
scatter = px.scatter(cat_merged_df, 
                     x='App', 
                     y='Installs',
                     title='Category Concentration',
                     size='App',
                     hover_name=cat_merged_df.index,
                     color='Installs'
) 

scatter.update_layout(xaxis_title="Number of Apps (Lower=More Concentrated)",
                      yaxis_title="Installs",
                      yaxis=dict(type='log'))

scatter.show()
```

```python
len(df_apps_clean.Genres.unique())
```

```python
df_apps_clean.Genres.value_counts().sort_values(ascending=True)[:5]
```

```python
stack = df_apps_clean.Genres.str.split(';', expand=True).stack()
print(f'We now have a single column with shape: {stack.shape}')
num_genres = stack.value_counts()
print(f'Number of genres: {len(num_genres)}')
```

```python
bar = px.bar(
        x = num_genres.index[:15], 
        y = num_genres.values[:15], 
        title='Top Genres',
        hover_name=num_genres.index[:15],
        color=num_genres.values[:15],
        color_continuous_scale='Agsunset'
)

bar.update_layout(xaxis_title='Genre',
                  yaxis_title='Number of Apps',
                  coloraxis_showscale=False)

bar.show()
```

```python
df_apps_clean.Type.value_counts()
```

```python
df_free_vs_paid = df_apps_clean.groupby(["Category", "Type"], 
                                      as_index=False).agg({'App': pd.Series.count})
df_free_vs_paid.sort_values('App')
```

```python
g_bar = px.bar(df_free_vs_paid, 
               x='Category', 
               y='App',
               title='Free vs Paid Apps by Category',
               color='Type', 
               barmode='group',)

g_bar.update_layout(xaxis_title='Category',
                    yaxis_title='Number of Apps',
                    xaxis={'categoryorder':'total descending'},
                    yaxis=dict(type='log'),
                    )

g_bar.show()
```

```python
box = px.box(df_apps_clean, 
             y='Installs',
             x='Type',
             color='Type',
             notched=True,
             points='all',
             title='How Many Downloads are Paid Apps Giving Up?'
)

box.update_layout(yaxis=dict(type='log'))

box.show()
```

```python
df_paid_apps = df_apps_clean[df_apps_clean['Type'] == 'Paid']
```
