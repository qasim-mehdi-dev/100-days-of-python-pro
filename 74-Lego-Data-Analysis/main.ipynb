```python
sets_df = pd.read_csv('data/sets.csv')
sets_df.head()
```

```python
sets_by_year = sets_df.groupby('year').count()
sets_by_year['set_num'].head()
```

```python
sets_by_year['set_num'].tail(10)
```

```python
import matplotlib.pyplot as plt

plt.plot(sets_by_year.index[:-2], sets_by_year['set_num'][:-2])
```

```python
themes_by_year = sets_df.groupby('year').agg({'theme_id': 'nunique'})
themes_by_year.head()
```

```python
plt.plot(themes_by_year.index[:-2], themes_by_year['theme_id'][:-2])
```

```python
ax1 = plt.gca() 
ax2 = ax1.twinx() 

ax1.plot(sets_by_year.index[:-2], sets_by_year['set_num'][:-2], color='g')
ax2.plot(themes_by_year.index[:-2], themes_by_year['theme_id'][:-2], color='b')

ax1.set_xlabel('Year')
ax1.set_ylabel('Number of Sets', color='g')
ax2.set_ylabel('Number of Themes', color='b')
```

```python
parts_by_year = sets_df.groupby('year').agg({'num_parts': 'mean'})
parts_by_year.head()
```

### Merging (i.e., Combining) DataFrames based on a Key


```python
themes_df.rename(columns={'id': 'theme_id'}, inplace=True)
themes_df.head()
```

```python
merged_df = pd.merge(sets_df, themes_df, on='theme_id')
merged_df.head()
```

```python
merged_df.rename(columns={'name_x': 'set_name', 'name_y': 'theme_name'}, inplace=True)
merged_df.head()
```

```python
merged_df['theme_name'].value_counts().head(10)
```

```python
top_themes = merged_df['theme_name'].value_counts().head(10)

plt.figure(figsize=(14,6))
plt.bar(top_themes.index, top_themes.values)

plt.title('Top 10 Most Popular LEGO Themes', fontsize=16)
plt.xlabel('Theme Name', fontsize=12)
plt.ylabel('Number of Sets', fontsize=12)
plt.xticks(rotation=45) 
```
